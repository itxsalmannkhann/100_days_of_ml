### Important pd.read_json() Parameters

### 1. orient
Defines the structure of JSON data.

```python
pd.read_json("train.json", orient="records")
```

### 2. typ
Decides whether to return a DataFrame or Series.

* frame → DataFrame (default).
* series → Series.

```python
pd.read_json("train.json", typ="series")
```

```python
pd.read_json("train.json", typ="frame")
```

### 3. dtype
Controls the data types of columns.

```python
pd.read_json("train.json", dtype={"id": "float64"})
```

The id column is converted from integer to float.

### 4. lines
Reads one JSON object per line.

```python
# without lines parameter

pd.read_json("students_lines.json")
```

```python
pd.read_json('students_lines.json', lines=True)
```

### 5. chunksize
Reads a large JSON file in smaller batches.

```python
for chunk in pd.read_json("students_lines.json", lines=True, chunksize=2):
    print('\n', chunk)
```
