### Load Required Libraries

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

### Load Dataset

```python
df = pd.read_csv("books-dataset.csv")
```

```python
print("Original Shape:", df.shape)
```

### Remove duplicate books

```python
df = df.drop_duplicates(subset="openlibrary_key")
```

### Remove records without titles

```python
df = df.dropna(subset=["book_title"])
```

### Remove empty titles

```python
df = df[df["book_title"].str.strip() != ""]
```

### Fill missing values

```python
df['author'] = df['author'].fillna("Unknown")
df['publisher'] = df['publisher'].fillna('Unkown')
```

### EDA (Exploratory Data Analysis)

#### Create visualizations

* A. Top 10 authors

```python
top_authors = df['author'].value_counts().head(10)

top_authors.plot(kind='bar', figsize=(10,5))

plt.title('Top 10 Authors')
plt.xlabel('Authors')
plt.ylabel('Number of Books')
plt.show()
```

* B. Books published by year

```python
years_counts = (df['publication_year'].dropna().astype(int).value_counts().sort_index())
```

```python
years_counts.plot(figsize=(12,5))
plt.title('Books Published Over the Years')
plt.xlabel('Publication Year')
plt.ylabel('Number of Books')
plt.show()
```

* C. Distribution of edition counts

```python
sns.histplot(
    df['edition_count'],
    bins=20
)

plt.title('Distribution of Book Editions')
plt.xlabel('Number of Editions')
plt.ylabel('Number of Books')
plt.show()
```
