## TMDB DATASET CREATION

### Making single large dataset

```python
all_movies = []

for i in range(1, 501):

    url = "https://api.themoviedb.org/3/movie/top_rated"

    params = {
        "api_key": "5944226186f0f852b516b2859ff7c7bb",
        "language": "en-US",
        "page": i
    }

    response = requests.get(url, params=params)
    data = response.json()

    if response.status_code != 200:
        print("API Error:", data)
        break

    if "results" not in data:
        print("Results key not found:", data)
        break

    all_movies.extend(data["results"])

df = pd.DataFrame(all_movies)
```

### Extract the selected cols

```python
df = df[['id', 'title', 'overview', 'original_language',
         'release_date', 'genre_ids', 'popularity',
         'vote_average', 'vote_count']]
```

### ADD New Columns to Data

```python
df['release_year'] = pd.to_datetime(df['release_date'], errors='coerce').dt.year
df['release_month'] = pd.to_datetime(df['release_date'], errors='coerce').dt.month
```

## Creating Books Dataset

```python
url = "https://openlibrary.org/search.json"

books = []

for i in range(1, 200):
    
    params = {
        "q": "python",
        "fields": "key,title,author_name,first_publish_year,edition_count,publisher",
        "limit": 100,
        "page": i
    }

    headers = {
        "User-Agent": "BooksDatasetProject/1.0 (your-email@gmail.com)"
    }

    response = requests.get(
        url,
        params=params,
        headers=headers,
        timeout=30
    )


    if response.status_code == 200:

        data = response.json()

        for book in data.get("docs", []):
            record = {
                "openlibrary_key": book.get("key"),
                "book_title": book.get("title"),
                "author": ", ".join(book.get("author_name", [])),
                "publication_year": book.get("first_publish_year"),
                "edition_count": book.get("edition_count"),
                "publisher": ", ".join(book.get("publisher", []))
            }

            books.append(record)

    else:
        print("API Error: {}".format(response.status_code))
        print(response.text)

# Convert to DataFrame
df = pd.DataFrame(books)
```
