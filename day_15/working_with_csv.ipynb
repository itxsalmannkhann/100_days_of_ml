# 3. Opening a csv file from an URL

```python
import requests
from io import StringIO


url = 'https://raw.githubusercontent.com/cs109/2014_data/master/countries.csv'
headers = {"User-Agent": "Mozilla/5.0 (Macintosh; Intel Mac OS X 10.14; rv:66.0) Gecko/20100101 Firefox/66.0"}
req = requests.get(url, headers=headers)
data = StringIO(req.text)

pd.read_csv(data)
```

# Parameters

### read_csv() Parameters

>### 1. Sep Parameter
Handles alternative file separators, such as \t for Tab-Separated Values (TSV)

```python
# without sep parameter

pd.read_csv('movie_titles_metadata.tsv')
```

```python
# with sep parameter

pd.read_csv('movie_titles_metadata.tsv', sep='\t')
```

> ### 2. names parameter
Supplies custom column names when the raw CSV lacks a header row

```python
# without names parameter

pd.read_csv('movie_titles_metadata.tsv', sep='\t')
```

```python
# with names parameter

pd.read_csv('movie_titles_metadata.tsv', sep='\t', names=['sno','name','release_year','rating','votes','genres'])
```

> ### 3. header parameter
Specifies which row to use as the header (e.g., header=0 or header=None)

```python
# without header parameter
pd.read_csv('test.csv')
```

```python
# with header paramter

pd.read_csv('test.csv', header=1)
```
