# 1. Categorical Data

* ## 1. Team distribution

```python
print(df2['Team 1'].unique().value_counts())
print('\nTotal Teams : ', df2['Team 1'].unique().value_counts().sum())

sns.countplot(data=df2, x='Team 1', order=df2['Team 1'].value_counts().index)
plt.xticks(rotation=90)
plt.title('Team 1 Column')
```

```python
sns.countplot(data=df2, x='Team 2', order=df2['Team 2'].value_counts().index)
plt.xticks(rotation=90)
plt.title('Team 2 Column')
plt.show()
```

```python
sns.countplot(data=df2, x='Winner', order=df2['Winner'].value_counts().index)

plt.xticks(rotation=90)
plt.title('Distribution of Match Winners')
plt.show()
```

* ## 2. Ground / Venue distribution

```python
plt.figure(figsize=(14, 35))
sns.countplot(data=df2, y='Ground', order=df2['Ground'].value_counts().index)
plt.yticks(fontsize=12)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10,20))
df2['Ground'].value_counts().head(15).plot(kind='pie', autopct='%.2f')
plt.title('15 Major Grounds')
```

* ## 3. Innings Visualization

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

sns.countplot(data=df2, x='Innings_Team1', ax=axes[0], order=['First', 'Second'])

axes[0].set_title('Team 1 Innings Distribution')
axes[0].set_xlabel('Innings')
axes[0].set_ylabel('Number of Matches')

sns.countplot(data=df2, x='Innings_Team2', ax=axes[1], order=['First', 'Second'])

axes[1].set_title('Team 2 Innings Distribution')
axes[1].set_xlabel('Innings')
axes[1].set_ylabel('Number of Matches')

plt.tight_layout()
plt.show()
```

The innings distribution analysis reveals that Team 1 and Team 2 have an equal distribution of batting positions. Each team batted first in 3,747 matches and second in 3,747 matches. This represents a balanced 50%-50% distribution, indicating that neither team has a frequency advantage in batting order within the dataset.
